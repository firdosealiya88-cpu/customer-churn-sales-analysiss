```python

import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

# 1. Load the data matrices
churn_df = pd.read_csv("churn_data.csv")

# 2. Select foundational numeric tracking features
# We focus on numbers first because text columns require wider variety to process safely
features = ['Tenure', 'MonthlyCharges', 'TotalCharges', 'SeniorCitizen']
X = churn_df[features]
y = churn_df['Churn']

# 3. Train the Random Forest intelligence model directly on this profile matrix
model = RandomForestClassifier(n_estimators=50, random_state=42)
model.fit(X, y)
print("🤖 AI Churn Predictive Model Trained Successfully!\n")

# 4. Create a built-in interactive manual customer evaluation function
def run_mobile_churn_test(tenure, monthly, total, senior_flag):
    # Construct a real-time single-row evaluation profile
    test_profile = pd.DataFrame([{
        'Tenure': tenure,
        'MonthlyCharges': monthly,
        'TotalCharges': total,
        'SeniorCitizen': senior_flag
    }])

    # Extract structural probability matrices
    probabilities = model.predict_proba(test_profile)[0]
    prediction = model.predict(test_profile)[0]

    # Output formatted metric readings
    print("--- 📲 Real-Time Account Risk Profile ---")
    print(f"• Probability of Retaining Customer: {probabilities[0]*100:.1f}%")
    print(f"• Probability of Customer Attrition: {probabilities[1]*100:.1f}%")
    print(f"• Risk Category Assessment: {'⚠️ HIGH CHURN RISK' if prediction == 1 else '✅ SECURE ACCOUNT'}")

# --- RUN A SIMULATION ---
# Parameters: tenure (months), monthly_charge, total_charge, senior_citizen (1=yes, 0=no)
run_mobile_churn_test(tenure=5, monthly=120, total=600, senior_flag=1)
```

```python
import pandas as pd
import numpy as np
import io
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, accuracy_score

print("🚀 Starting the consolidated Customer Analytics Pipeline...\n")

# =====================================================================
# STEP 1: EMBEDDED REVENUE & CHURN DATA LOGS
# =====================================================================
sales_raw_data = """Date,Product,Quantity,Price,Customer_ID,Region,Total_Sales
2024-01-01,Phone,7,37300,CUST001,East,261100
2024-01-02,Headphones,4,15406,CUST002,North,61624
2024-01-03,Phone,2,21746,CUST003,West,43492
2024-01-04,Headphones,1,30895,CUST004,East,30895
2024-01-05,Laptop,8,39835,CUST005,North,318680
2024-01-06,Laptop,7,40420,CUST006,South,282940
2024-01-07,Laptop,9,40430,CUST007,South,363870
2024-01-08,Laptop,7,7262,CUST008,West,50834
2024-01-09,Tablet,3,32791,CUST009,North,98373
2024-01-10,Laptop,4,45023,CUST010,West,180092
2024-01-11,Monitor,8,6160,CUST011,West,49280
2024-01-12,Phone,2,13233,CUST012,West,26466
2024-01-13,Phone,8,20655,CUST013,East,165240
2024-01-14,Headphones,4,4195,CUST014,East,16780
2024-01-15,Headphones,2,34447,CUST015,West,68894
2024-01-16,Phone,9,41548,CUST016,South,373932
2024-01-17,Phone,5,13488,CUST017,South,67440
2024-01-18,Tablet,4,19322,CUST018,North,77288
2024-01-19,Tablet,1,44045,CUST019,West,44045
2024-01-20,Laptop,8,41749,CUST020,North,333992"""

churn_raw_data = """CustomerID,Tenure,MonthlyCharges,TotalCharges,Contract,PaymentMethod,PaperlessBilling,SeniorCitizen,Churn
C00001,6,64,1540,One year,Credit Card,No,1,0
C00002,21,113,1753,Month-to-month,Electronic Check,Yes,1,0
C00012,7,110,7282,Month-to-month,Electronic Check,No,1,1
C00014,8,158,6194,Month-to-month,Credit Card,No,1,1
C00005,16,185,1023,One year,Electronic Check,No,1,0
C00019,7,156,5389,One year,Bank Transfer,Yes,0,1
C00007,66,99,4178,Two year,Bank Transfer,No,1,0
C00074,2,152,2038,Month-to-month,Bank Transfer,No,0,1
C00009,64,124,7973,Month-to-month,Credit Card,No,1,0
C00107,3,59,7111,Month-to-month,Credit Card,Yes,1,1
C00025,25,193,7565,Month-to-month,Bank Transfer,No,0,0
C00032,10,196,4825,Month-to-month,Credit Card,Yes,0,1
C00038,62,166,6225,Two year,Electronic Check,Yes,1,0
C00055,9,157,7022,Month-to-month,Bank Transfer,No,0,1
C00068,62,62,3016,Two year,Credit Card,Yes,0,0
C00125,4,177,950,Month-to-month,Credit Card,Yes,0,1"""

sales_df = pd.read_csv(io.StringIO(sales_raw_data))
churn_df = pd.read_csv(io.StringIO(churn_raw_data))

# =====================================================================
# STEP 2: TRANSACTIONAL BI DASHBOARD & PROFITABILITY TRACKING
# =====================================================================
print("--- 💸 Part 1: Sales Performance Dashboard Summary ---")
product_revenue = sales_df.groupby('Product')['Total_Sales'].sum().sort_values(ascending=False)
for product_name, revenue_amount in product_revenue.items():
    print(f"• {product_name}: ${revenue_amount:,}")

# =====================================================================
# STEP 3: MACHINE LEARNING PIPELINE PREPROCESSING
# =====================================================================
X = churn_df[['Tenure', 'MonthlyCharges', 'TotalCharges', 'Contract', 'PaymentMethod', 'PaperlessBilling', 'SeniorCitizen']]
y = churn_df['Churn']

numeric_features = ['Tenure', 'MonthlyCharges', 'TotalCharges']
categorical_features = ['Contract', 'PaymentMethod', 'PaperlessBilling', 'SeniorCitizen']

data_transformer = ColumnTransformer(
    transformers=[
        ('num_scaler', StandardScaler(), numeric_features),
        ('cat_encoder', OneHotEncoder(handle_unknown='ignore', drop='first'), categorical_features)
    ])

# Separate data into training subsets (75%) and test validation slices (25%)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42, stratify=y)

X_train_clean = data_transformer.fit_transform(X_train)
X_test_clean = data_transformer.transform(X_test)

# =====================================================================
# STEP 4: MODEL TRAINING & METRIC ASSESSMENTS
# =====================================================================
random_forest_model = RandomForestClassifier(n_estimators=100, random_state=42)
random_forest_model.fit(X_train_clean, y_train)

predictions = random_forest_model.predict(X_test_clean)
print("\n--- 🤖 Part 2: Machine Learning Model Evaluation Matrix ---")
print(f"Overall Classification Accuracy Score: {accuracy_score(y_test, predictions) * 100:.1f}%")

# =====================================================================
# STEP 5: VISUAL DATA VIZ GENERATION
# =====================================================================
print("\n📊 Part 3: Generating Visualizations Charts...")
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Plot A: Total Accumulated Income Bar Plot
sns.barplot(x=product_revenue.values, y=product_revenue.index, ax=axes[0], hue=product_revenue.index, legend=False, palette="viridis")
axes[0].set_title("Total Sales Revenue by Product Segment")
axes[0].set_xlabel("Total Sales Amount ($)")
axes[0].set_ylabel("Product Category Names")

# Plot B: Customer Base Churn Counting Map
sns.countplot(data=churn_df, x='Churn', ax=axes[1], hue='Churn', legend=False, palette="Set2")
axes[1].set_title("Subscriber Retention Distribution")
axes[1].set_xticks([0, 1])
axes[1].set_xticklabels(['Retained Base (0)', 'Churned Base (1)'])
axes[1].set_ylabel("Number of Customer Accounts")

plt.tight_layout()
plt.show()

# =====================================================================
# STEP 6: CORRECTED DYNAMIC RUNTIME USER TESTING PATTERN
# =====================================================================
def dynamic_account_evaluator(tenure, monthly, total, contract, payment, paperless, senior):
    # Establish single test item row matching baseline features structure
    custom_profile = pd.DataFrame([{
        'Tenure': tenure,
        'MonthlyCharges': monthly,
        'TotalCharges': total,
        'Contract': contract,
        'PaymentMethod': payment,
        'PaperlessBilling': paperless,
        'SeniorCitizen': senior
    }])

    # Process properties using preprocessor configuration
    cleaned_input = data_transformer.transform(custom_profile)
    probability_matrix = random_forest_model.predict_proba(cleaned_input)[0]
    final_prediction = random_forest_model.predict(cleaned_input)[0]

    print("\n--- 📲 Part 4: Real-Time Account Risk Output Profile ---")
    print(f"• Probability of Retaining Customer: {probability_matrix[0]*100:.1f}%")
    print(f"• Probability of Customer Attrition: {probability_matrix[1]*100:.1f}%")
    print(f"• Final System Judgment: {'⚠️ ALERT: HIGH CHURN RISK' if final_prediction == 1 else '✅ SECURE ACCOUNT'}")

# --- EXECUTE STABLE VERIFIED RUNTIME TESTING ---
dynamic_account_evaluator(
    tenure=3,
    monthly=150,
    total=450,
    contract='Month-to-month',
    payment='Electronic Check',  # Uniform keyword fix resolved
    paperless='Yes',
    senior=1
)
```
